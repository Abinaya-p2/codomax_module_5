```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/diabetes_012_health_indicators_BRFSS2015.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

display(df.describe())
```

```python
print("Original rows:", len(df))
print("Duplicate rows:", df.duplicated().sum())

df_clean = df.drop_duplicates().copy()

print("Rows after removing duplicates:", len(df_clean))
print("Remaining duplicate rows:", df_clean.duplicated().sum())
```

```python
print("Diabetes Category Counts")
print(df_clean["Diabetes_012"].value_counts().sort_index())

print("\nPercentage of each category:")
print(
    (df_clean["Diabetes_012"].value_counts(normalize=True)
     .sort_index() * 100).round(2)
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))
sns.countplot(data=df_clean, x="Diabetes_012")

plt.title("Distribution of Diabetes Categories")
plt.xlabel("Diabetes Category (0 = No, 1 = Prediabetes, 2 = Diabetes)")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python
avg_bmi = df_clean.groupby("Age")["BMI"].mean()

avg_bmi.plot(kind="line", marker="o", figsize=(8, 5))

plt.title("Average BMI by Age Category")
plt.xlabel("Age Category")
plt.ylabel("Average BMI")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_clean,
    x="PhysActivity",
    hue="Diabetes_012"
)

plt.title("Physical Activity by Diabetes Category")
plt.xlabel("Physical Activity (0 = No, 1 = Yes)")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python
print("HEALTHCARE DATA ANALYSIS FINDINGS")
print("=================================")

print("\n1. Diabetes category counts:")
print(df_clean["Diabetes_012"].value_counts().sort_index())

print("\n2. Average BMI by category:")
print(df_clean.groupby("Diabetes_012")["BMI"].mean().round(2))

print("\n3. High blood pressure counts:")
print(df_clean["HighBP"].value_counts())

print("\n4. Physical activity counts:")
print(df_clean["PhysActivity"].value_counts())

print("\n5. Average general health rating by category:")
print(df_clean.groupby("Diabetes_012")["GenHlth"].mean().round(2))
```

```python
df_clean.to_csv(
    "diabetes_health_indicators_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
